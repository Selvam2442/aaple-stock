```python
import pandas as pd
import matplotlib as plt
import seaborn as sns


df = pd.read_csv("/content/shopify_stock.csv")
df["date"] = pd.to_datetime(df["date"])
```

```python
df.head()
```

```python
df.dtypes
```

```python
plt.pyplot.figure(figsize=(10, 5))

plt.pyplot.plot(df["date"], df["close"])

plt.pyplot.title("Shopify Closing Price")
plt.pyplot.xlabel("Date")
plt.pyplot.ylabel("Closing Price")

plt.pyplot.show()
```

```python
df["MA20"] = df["close"].rolling(20).mean()
df["MA50"] = df["close"].rolling(50).mean()

plt.pyplot.figure(figsize=(10, 5))

plt.pyplot.plot(df["date"], df["close"], label="Closing Price")
plt.pyplot.plot(df["date"], df["MA20"], label="20-Day MA")
plt.pyplot.plot(df["date"], df["MA50"], label="50-Day MA")

plt.pyplot.title("Shopify Moving Averages")
plt.pyplot.xlabel("Date")
plt.pyplot.ylabel("Price")
plt.pyplot.legend()

plt.pyplot.show()
```

```python
df["Daily_Return"] = df["close"].pct_change()

print(df[["date", "close", "Daily_Return"]].head())
```

```python
plt.pyplot.figure(figsize=(10, 5))

plt.pyplot.hist(df["Daily_Return"].dropna(), bins=30)

plt.pyplot.title("Shopify Daily Returns Histogram")
plt.pyplot.xlabel("Daily Return")
plt.pyplot.ylabel("Frequency")

plt.pyplot.show()
```

```python
plt.pyplot.figure(figsize=(10, 5))

sns.kdeplot(df["Daily_Return"].dropna(), fill=True)

plt.pyplot.title("Shopify Daily Returns KDE Plot")
plt.pyplot.xlabel("Daily Return")
plt.pyplot.ylabel("Density")

plt.pyplot.show()
```
